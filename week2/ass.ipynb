```python
# load your YAML world
env = irsim.make("mrw.yaml")

robot = env.robot  # get the first robot

# define circle path
center = np.array([25, 25])
radius = 10
num_points = 100
theta_list = np.linspace(0, 2*np.pi, num_points)

path = [(center[0]+radius*np.cos(t), center[1]+radius*np.sin(t)) for t in theta_list]

for x, y in path:
    robot.state[0] = x   # x
    robot.state[1] = y   # y
    
    env.step()
    env.render()
```

```python
env = irsim.make("mrw.yaml")
robots = env.robot  # RobotGroup with 3 robots

center = np.array([25, 25])
radius = 10
num_points = 200
theta_list = np.linspace(0, 2*np.pi, num_points)

# circle path
path = [(center[0] + radius*np.cos(t), center[1] + radius*np.sin(t), 0) for t in theta_list]

# offsets so robots are spaced evenly
offsets = [0, num_points // 3, 2 * num_points // 3]

for step in range(600):  # run 3 laps
    for i in range(3):  # loop over 3 robots
        idx = (step + offsets[i]) % num_points
        robots[i].goal = [path[idx]]  # assign next goal on circle
    
    env.step()
    env.render(0.05)

env.end(3)
```
