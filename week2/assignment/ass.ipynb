# Making the robot follows a circle path 

```python
# load your YAML world
env = irsim.make("world.yaml")

robot = env.robot  # get the first robot

# define circle path
center = np.array([25, 25])
radius = 10
num_points = 100
theta_list = np.linspace(0, 2*np.pi, num_points)

path = [(center[0]+radius*np.cos(t), center[1]+radius*np.sin(t)) for t in theta_list]

for x, y in path:
    robot.state[0] = x   # x
    robot.state[1] = y   # y
    
    env.step()
    env.render()
```
